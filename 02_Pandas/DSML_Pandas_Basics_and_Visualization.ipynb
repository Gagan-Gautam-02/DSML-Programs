### Completing: adding/modifying data, missing values, sorting & filtering (Iris)

```python
iris = data.copy()
iris['PetalArea'] = iris['PetalLengthCm'] * iris['PetalWidthCm']            # add
iris['Species'] = iris['Species'].str.replace('Iris-', '')                  # modify
iris.loc[iris['SepalLengthCm'] > 7.5, 'Size'] = 'Large'
iris['Size'] = iris['Size'].fillna('Normal')                                 # fill missing values
iris.head()
```

### Correlation and heatmap

```python
num_cols = ['Rating', 'Founded', 'avg_salary', 'min_salary', 'max_salary', 'age', 'python_yn', 'spark', 'aws', 'excel']
plt.figure(figsize=(9, 7))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation heatmap: job postings')
plt.show()
```

```python
plt.figure(figsize=(5, 4))
sns.heatmap(df2[['Experience', 'Salary']].corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Experience vs Salary correlation')
plt.show()
```

### Location vs Job title: countplot

```python
plt.figure(figsize=(11, 5))
sns.countplot(data=df2, x='Location', hue='JobRole')
plt.title('Number of employees per job role in each location')
plt.xticks(rotation=30)
plt.legend(bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
