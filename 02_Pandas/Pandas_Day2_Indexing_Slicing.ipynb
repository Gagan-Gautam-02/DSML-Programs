## 2. Adding and modifying data

```python
import numpy as np

retail = df.copy()
retail['TotalPrice'] = retail['Quantity'] * retail['UnitPrice']          # new column from existing ones
retail['InvoiceDate'] = pd.to_datetime(retail['InvoiceDate'], format='%m/%d/%Y %H:%M')
retail['Month'] = retail['InvoiceDate'].dt.to_period('M')
retail['IsCancelled'] = retail['InvoiceNo'].astype(str).str.startswith('C')
retail[['InvoiceNo', 'Quantity', 'UnitPrice', 'TotalPrice', 'Month', 'IsCancelled']].head()
```

```python
retail.loc[retail['StockCode'] == 'POST', 'Description'] = 'POSTAGE CHARGE'   # conditional modify
retail['Description'] = retail['Description'].str.strip()
retail = retail.rename(columns={'UnitPrice': 'Price'})
retail.loc[len(retail)] = ['999999', 'TEST', 'MANUAL TEST ROW', 1, pd.Timestamp('2011-12-10'), 0.0, np.nan,
                           'United Kingdom', 0.0, pd.Period('2011-12', 'M'), False]            # add a row
retail.tail(2)
```

```python
retail = retail.drop(index=retail.index[-1])            # delete that test row
retail = retail.drop(columns=['IsCancelled'])           # delete a column
retail.shape
```

## 3. Handling missing values

```python
retail.isnull().sum()
```

```python
print('% missing CustomerID:', round(retail['CustomerID'].isnull().mean() * 100, 2))
```

```python
retail['Description'] = retail['Description'].fillna('UNKNOWN')          # fill text
with_customer = retail.dropna(subset=['CustomerID']).copy()               # drop rows missing the key
with_customer['CustomerID'] = with_customer['CustomerID'].astype(int)
print('rows before:', len(retail), '| after dropping missing CustomerID:', len(with_customer))
with_customer.isnull().sum()
```

## 4. Sorting and filtering

```python
with_customer.sort_values(by='TotalPrice', ascending=False).head()
```

```python
with_customer.sort_values(by=['Country', 'InvoiceDate'], ascending=[True, False]).head()
```

```python
with_customer[with_customer['Quantity'] < 0].head()          # cancelled orders
```

```python
with_customer[(with_customer['Country'] == 'France') & (with_customer['Quantity'] >= 12)].head()
```

```python
with_customer[with_customer['Country'].isin(['Germany', 'EIRE'])].head()
```

```python
with_customer[with_customer['Description'].str.contains('HEART')].head()
```

```python
with_customer.query('Price > 10 and Quantity > 0').head()
```

```python
# Putting it together: revenue by country from valid (non-cancelled) orders
valid = with_customer[with_customer['Quantity'] > 0]
valid.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False).head(5).round(2)
```
