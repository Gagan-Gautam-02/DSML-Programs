## 5. Adding / modifying / deleting columns

```python
students = df.copy()                                   # never modify the raw data
students['Total'] = students[['Maths', 'Physics', 'Chemistry']].sum(axis=1, min_count=3)
students['Percentage'] = (students['Total'] / 300 * 100).round(2)
students.insert(2, 'Batch', 2024)                       # insert at a chosen position
students.head()
```

```python
students.loc[students['Student_ID'] == 'S001', 'Study_Hours'] = 13     # modify one value
students = students.rename(columns={'Study_Hours': 'Weekly_Study_Hours'})
students = students.drop(columns=['Batch'])
students.columns.tolist()
```

## 6. Handling missing values

```python
students.isnull().sum()
```

```python
students[students.isnull().any(axis=1)]
```

```python
# Strategy per column
students['Age'] = students['Age'].fillna(students['Age'].median())
students['Attendance'] = students['Attendance'].fillna(students['Attendance'].mean().round(1))
students['City'] = students['City'].fillna(students['City'].mode()[0])
# Marks: fill with that department's median, which is fairer than a global value
for col in ['Maths', 'Physics']:
    students[col] = students[col].fillna(students.groupby('Department')[col].transform('median'))
students['Total'] = students[['Maths', 'Physics', 'Chemistry']].sum(axis=1)
students['Percentage'] = (students['Total'] / 300 * 100).round(2)
students.isnull().sum().sum()
```

```python
# The alternative: drop rows instead of filling
print('rows before dropna:', len(df), '| after:', len(df.dropna()))
```

## 7. Duplicates

```python
print('duplicate rows:', students.duplicated().sum())
students[students.duplicated(keep=False)].sort_values('Student_ID')
```

```python
students = students.drop_duplicates().reset_index(drop=True)
students.shape
```

## 8. Cleaning text columns
The same city is spelled in different ways, which would split it into separate groups.

```python
students['City'].value_counts()
```

```python
students['City'] = students['City'].str.strip().str.title()
students['Name'] = students['Name'].str.strip()
students['First_Name'] = students['Name'].str.split().str[0]
students['City'].value_counts()
```

## 10. GroupBy & aggregation

```python
students.groupby('Department')['Percentage'].mean().round(2).sort_values(ascending=False)
```

```python
students.groupby('Department').agg(
    students=('Student_ID', 'count'),
    avg_hours=('Weekly_Study_Hours', 'mean'),
    avg_attendance=('Attendance', 'mean'),
    avg_percentage=('Percentage', 'mean'),
    topper_total=('Total', 'max'),
).round(2)
```

```python
students.groupby(['Department', 'Gender'])['Percentage'].mean().unstack().round(2)
```

## 11. Apply / map

```python
def grade(p):
    if p >= 85: return 'A'
    if p >= 70: return 'B'
    if p >= 55: return 'C'
    if p >= 40: return 'D'
    return 'F'

students['Grade'] = students['Percentage'].apply(grade)
students['Gender_Full'] = students['Gender'].map({'M': 'Male', 'F': 'Female'})
students['Hours_Band'] = pd.cut(students['Weekly_Study_Hours'], bins=[0, 6, 12, 25], labels=['Low', 'Medium', 'High'])
students[['Name', 'Percentage', 'Grade', 'Gender_Full', 'Hours_Band']].head()
```

```python
students['Grade'].value_counts().sort_index()
```

## Key takeaways
- Always start with `head()`, `info()`, `describe()` and `isnull().sum()`.
- `loc` includes the end label, while `iloc` excludes the end position.
- Choose imputation per column: median for skewed numbers, mode for categories, and group-wise values when groups differ.
- Clean text (`str.strip().str.title()`) **before** grouping, or one city becomes several.
- `groupby` + `agg` is the fastest way to answer "which X has the highest Y?"
- Work on a `.copy()` so the raw data is never modified by accident.
