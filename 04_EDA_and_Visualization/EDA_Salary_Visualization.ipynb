```python
#Univariate Analysis
#Distribution of employees by salary
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data['Salary'], bins = 10, kde= True)
plt.title('Distribution of Employees by Salary')
plt.xlabel('Salary')
plt.ylabel('Number of Employes')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data['Experience'], bins = 10, kde= True)
plt.title('Experience Distribution')
plt.xlabel('Years of Expirence')
plt.ylabel('Number of Employes')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=data, x = "JobRole")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=data, x = "JobRole", y = "Salary")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=data, x = "Education", y = "Salary")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(data=data, x = "Experience", y = "Salary")
plt.show()
```
