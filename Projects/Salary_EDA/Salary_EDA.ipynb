# ***EDA USING SALARY DATASET***

```python
#import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load the dataset
df=pd.read_csv("datasets/salary_dataset.csv")
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
#checking for missing values

df.isnull().sum()
```

```python
#Univariate analysis
#Distribution of employess by salary

plt.figure(figsize=(8,5))
sns.histplot(df['Salary'], bins=10, kde=True)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Numbe of Employees")
plt.show()
```

```python
#Experience Distribution

plt.figure(figsize=(8,5))
sns.histplot(df['Experience'], bins=10, kde=True)
plt.title("Experience Distribution")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#No. of employees by Job Role

plt.figure(figsize=(8,5))
sns.countplot(data=df,x="JobRole")

plt.title("Number of Employees by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

```python
#Average salary by jobrole
#Bivariate analysis (including two variables)

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="JobRole", y="Salary")
plt.title("Avg. salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Avg. Salary")
plt.xticks(rotation=45)
plt.show()
```

```python
#Salary by Education

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Education", y="Salary")
plt.title("Avg. salary by Education")
plt.xlabel("Education")
plt.ylabel("Avg. Salary")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Location", y="Salary")
plt.title("Avg. salary by Location")
plt.xlabel("Location")
plt.ylabel("Avg. Salary")
plt.show()
```

```python
#Experience vs Salary

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Experience", y="Salary")

plt.title("Experience v/s Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
#Experience vs Salary with regression line

plt.figure(figsize=(8,5))
sns.regplot(data=df, x="Experience", y="Salary")

plt.title("Experience v/s Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
#Salary Distribution with Education

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Education", y="Salary")

plt.title("Education v/s Salary")
plt.xlabel("Education")
plt.ylabel("Salary")
plt.show()
```
